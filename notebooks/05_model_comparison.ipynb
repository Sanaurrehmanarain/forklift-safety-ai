# Goal: Train multiple architectures and compare their speed and accuracy.

Model Comparison: YOLO11-Nano vs YOLO11-Small

This notebook trains two different models on our dataset to compare mAP and inference speed.

```python
import os
import time
from ultralytics import YOLO
import pandas as pd
import matplotlib.pyplot as plt

YAML_PATH = "../data/dataset.yaml"
PROJECT_DIR = "../trained_models"

models_to_test = {"YOLO11-Nano": "yolo11n.pt", "YOLO11-Small": "yolo11s.pt"}
results_data = []

for name, weights in models_to_test.items():
    print(f"\nTraining {name}...")
    model = YOLO(weights)
    
    # Train
    train_results = model.train(data=YAML_PATH, epochs=10, imgsz=640, project=PROJECT_DIR, name=name)
    
    # Measure Inference Speed
    start = time.time()
    model.predict("../data/processed/images/val", verbose=False)
    end = time.time()
    
    results_data.append({
        "Model": name,
        "mAP@50": train_results.box.map50,
        "Inference_Time_sec": end - start
    })

# Comparison Results

df = pd.DataFrame(results_data)
print(df)

df.plot(x="Model", y=["mAP@50", "Inference_Time_sec"], kind="bar", subplots=True, layout=(1,2), figsize=(12,4))
plt.show()
```
