# Goal: Generate deep statistical metrics (Confusion Matrix, Precision-Recall curves) on the validation set.

Model Evaluation

Generating rigorous statistical metrics for our best trained model.

```python
from ultralytics import YOLO
from IPython.display import Image, display
import os

# Load the best model
model = YOLO("../trained_models/yolo_worker_safety/weights/best.pt")

# Run validation. This automatically generates confusion matrices and PR curves.
metrics = model.val(data="../data/dataset.yaml", project="../reports", name="val_evaluation")

print(f"Mean Average Precision @50: {metrics.box.map50:.3f}")
print(f"Mean Average Precision @50-95: {metrics.box.map:.3f}")

# Confusion Matrix

cm_path = "../reports/val_evaluation/confusion_matrix.png"
if os.path.exists(cm_path):
    display(Image(filename=cm_path))
else:
    print("Confusion matrix not generated.")
```
