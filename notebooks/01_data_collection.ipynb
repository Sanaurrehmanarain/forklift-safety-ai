# Dataset Collection

```python
import cv2
import os
import urllib.request


VIDEO_DIR = "../data/videos"
RAW_DIR = "../data/raw"

os.makedirs(VIDEO_DIR, exist_ok=True)
os.makedirs(RAW_DIR, exist_ok=True)

# Download a sample warehouse forklift/worker safety video for testing
video_url = "https://github.com/intel-iot-devkit/sample-videos/raw/master/worker-zone-detection.mp4"
video_path = os.path.join(VIDEO_DIR, "sample_warehouse.mp4")

if not os.path.exists(video_path):
    print("Downloading sample video...")
    urllib.request.urlretrieve(video_url, video_path)
    print("Download complete!")

# Video processing setup
cap = cv2.VideoCapture(video_path)
fps = round(cap.get(cv2.CAP_PROP_FPS))
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
duration = total_frames / fps if fps > 0 else 0

print(f"--- Video Statistics ---")
print(f"Total Frames: {total_frames}")
print(f"Original FPS: {fps}")
print(f"Duration: {duration:.2f} seconds\n")

# Extract frames
print("Extracting frames...")
frame_id = 0
saved_count = 0

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break
        
    # Save 1 frame every 'fps' frames (approx 1 image per second of video)
    if frame_id % fps == 0:
        frame_name = os.path.join(RAW_DIR, f"frame_{saved_count:04d}.jpg")
        cv2.imwrite(frame_name, frame)
        saved_count += 1
        
    frame_id += 1

cap.release()
print(f"Extraction complete! Saved {saved_count} unique frames to '{RAW_DIR}'.")
```
