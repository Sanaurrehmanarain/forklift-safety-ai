# Goal: Convert the PyTorch (`.pt`) model into deployment-optimized formats (ONNX, OpenVINO).

Model Optimization (Exporting)

Exporting the model to ONNX for cross-platform deployment and OpenVINO for Intel/AMD CPU acceleration.

```python
from ultralytics import YOLO

model = YOLO("../trained_models/yolo_worker_safety/weights/best.pt")

print("Exporting to ONNX format...")
# ONNX is great for general deployment (C++, C#, Node.js)
onnx_path = model.export(format="onnx", half=False) 

print("Exporting to OpenVINO format...")
# OpenVINO drastically speeds up inference on local CPUs without a GPU
openvino_path = model.export(format="openvino", half=True) # FP16 Quantization

print("Optimization complete. Ready for edge deployment.")
```
