# Tracking the Workers

```python
import cv2
import os
from ultralytics import YOLO

# Define paths
VIDEO_PATH = "../data/videos/sample_warehouse.mp4"
OUTPUT_DIR = "../outputs"
os.makedirs(OUTPUT_DIR, exist_ok=True)
OUTPUT_PATH = os.path.join(OUTPUT_DIR, "tracked_warehouse.mp4")

# Locate your custom trained weights from the log output
custom_weights = "runs/trained_models/yolo_worker_safety/weights/best.pt"

if os.path.exists(custom_weights):
    print(f"Loading your custom trained model from {custom_weights}")
    model = YOLO(custom_weights)
else:
    print("Custom weights not found. Falling back to base YOLO11n.")
    model = YOLO("yolo11n.pt")

print("Starting video tracking... Processing frame-by-frame on CPU.")

# Open the original video
cap = cv2.VideoCapture(VIDEO_PATH)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = int(cap.get(cv2.CAP_PROP_FPS))

# Setup VideoWriter to save the output
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(OUTPUT_PATH, fourcc, fps, (width, height))

frame_count = 0
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

while cap.isOpened():
    success, frame = cap.read()
    if not success:
        break
    
    # Run YOLO tracking (persist=True keeps track of IDs across frames)
    # set verbose=False so it doesn't spam your terminal
    results = model.track(frame, persist=True, verbose=False)
    
    # Visualize the results (draws boxes and IDs on the frame)
    annotated_frame = results[0].plot()
    
    # Write the annotated frame to new output video
    out.write(annotated_frame)
    
    frame_count += 1
    if frame_count % 150 == 0:
        print(f"Processed {frame_count}/{total_frames} frames...")

cap.release()
out.release()
print(f"\nTracking complete! AI-annotated video saved to: {OUTPUT_PATH}")
```
