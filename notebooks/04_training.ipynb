# Training the Safety Monitor

```python
import os
import torch
from ultralytics import YOLO
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

YAML_PATH = "../data/dataset.yaml"
PROJECT_DIR = "../trained_models"
RUN_NAME = "yolo_worker_safety"

os.makedirs(PROJECT_DIR, exist_ok=True)

if torch.cuda.is_available():
    device_to_use = "cuda"
    print("NVIDIA GPU detected! Training will be fast.")
elif hasattr(torch.backends, "mps") and torch.backends.mps.is_available():
    device_to_use = "mps"
    print("Apple Silicon GPU detected! Training on MPS.")
else:
    device_to_use = "cpu"
    print("No compatible GPU found. Training on CPU. (This will be slower, but it will work!)")

print("Initializing YOLO11 Model...")
model = YOLO("yolo11n.pt") 

print("\nStarting Training Phase...")
print(f"Parameters: Epochs=10, Batch=8, ImgSz=640, Device={device_to_use}")

# Train the model
results = model.train(
    data=YAML_PATH,
    epochs=10,
    imgsz=640,
    batch=8,
    project=PROJECT_DIR,
    name=RUN_NAME,
    exist_ok=True,       # Overwrite if we run this cell multiple times
    patience=5,          # Early stopping
    device=device_to_use # Pass our clean local hardware string here
)

print(f"\nTraining Complete! Model weights saved to: {PROJECT_DIR}/{RUN_NAME}/weights/")

# Visualize the training results automatically
results_img_path = f"{PROJECT_DIR}/{RUN_NAME}/results.png"

if os.path.exists(results_img_path):
    print("\nDisplaying Training Metrics...")
    plt.figure(figsize=(12, 8))
    img = mpimg.imread(results_img_path)
    plt.imshow(img)
    plt.axis('off')
    plt.show()
else:
    print(f"Could not find results chart at {results_img_path}")
```

If you look closely at your logs, Ultralytics auto-generated a `runs/` folder inside the `notebooks/` directory because of how it resolves relative paths. It saved the chart and model weights here:
`notebooks\runs\trained_models\yolo_worker_safety\weights\best.pt`

This is completely fine. The weights are saved, and that is all we need for the next step.
