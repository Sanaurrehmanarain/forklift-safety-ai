# Goal: Visually inspect where the model fails (False Positives/Negatives).

Error Analysis

Saving low-confidence detections and crops to manually inspect edge cases (e.g., motion blur, occlusion).

```python
from ultralytics import YOLO

model = YOLO("../trained_models/yolo_worker_safety/weights/best.pt")

# Predict on validation set, saving crops of detections and setting a high threshold 
# to see what it *misses* (False Negatives), or a low threshold to see what it *hallucinates* (False Positives).
print("Running inference and saving edge-cases...")

results = model.predict(
    source="../data/processed/images/val", 
    conf=0.20,             # Low confidence to inspect potential false positives
    save=True,             # Save annotated images
    save_crop=True,        # Save cropped images of the detected workers
    project="../reports",
    name="error_analysis"
)

print("Review the '../reports/error_analysis' folder for visual inspection.")
```
